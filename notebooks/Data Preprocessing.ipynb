```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler

from ucimlrepo import fetch_ucirepo
```

```python
parkinsons = fetch_ucirepo(id=174) 

X = parkinsons.data.features
y = parkinsons.data.targets

df = pd.concat([X, y], axis=1)
df.head()
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(X.corr(), annot=False, cmap='coolwarm')
plt.title("Correlation Heatmap of Parkinson's Features")
plt.show()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
print("Data successfully scaled!")
```

```python
import warnings
warnings.filterwarnings('ignore')
```

**K-Means**

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
```

```python
# Elbow Method (Find Optimal k)
k_optimal = 3  

inertia = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled_df)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K_range, inertia, 'bo-', linewidth=2, markersize=8, markeredgecolor='black')
plt.axvline(x=k_optimal, color='red', linestyle='--', alpha=0.7, label=f'Optimal k={k_optimal}')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia (WCSS)')
plt.title('Elbow Method for Optimal k')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Silhouette Analysis
sil_scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_scaled_df)
    sil_scores.append(silhouette_score(X_scaled_df, labels))

plt.figure(figsize=(8,5))
plt.plot(range(2,11), sil_scores, marker='o')
plt.xlabel('k')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Analysis')
plt.show()
```

```python
# Fit Final K-Means Model
k_optimal = 3
kmeans = KMeans(n_clusters=k_optimal, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X_scaled_df)
df['Cluster'] = clusters
```

```python
# Evaluate Clustering
sil = silhouette_score(X_scaled_df, clusters)
dbi = davies_bouldin_score(X_scaled_df, clusters)
chi = calinski_harabasz_score(X_scaled_df, clusters)

print(f"Silhouette Score: {sil}")
print(f"Davies-Bouldin Index: {dbi}")
print(f"Calinski-Harabasz Index: {chi}")
```

The K-means clustering model identified moderate cluster structures within the dataset. The silhouette score of 0.277 indicated weak to moderate separation between clusters, suggesting some overlap among data points. The Davies–Bouldin Index of 1.205 showed moderate cluster compactness and similarity, while the Calinski–Harabasz Index of 104.16 suggested a reasonable degree of cluster separation. Overall, the results indicate that K-means was able to capture some underlying structure in the data, although the clusters were not strongly distinct.

```python
# PCA for Visualization
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled_df)

plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0], X_pca[:,1], c=clusters, cmap='viridis')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.title('K-Means Clusters (PCA Projection)')
plt.colorbar()
plt.show()
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled_df)
```

```python
explained_variance = pca.explained_variance_ratio_

print("Variance explained by PC1:", explained_variance[0])
print("Variance explained by PC2:", explained_variance[1])
print("Total variance explained:", explained_variance.sum())
```

The first principal component explained 59.25% of the total variance, while the second principal component explained 11.31%. Together, the two components retained approximately 70.56% of the dataset information, indicating that PCA successfully preserved most of the important structure of the original data while reducing dimensionality.

```python
kmeans_pca = KMeans(
    n_clusters=k_optimal,
    init='k-means++',
    random_state=42
)

clusters_pca = kmeans_pca.fit_predict(X_pca)

# Plot cluster centers in PCA space
plt.figure(figsize=(10, 8))
scatter = plt.scatter(X_pca[:,0], X_pca[:,1], c=clusters_pca, cmap='viridis', 
                      alpha=0.6, s=50, edgecolors='black', linewidth=0.5)

# Plot cluster centers
cluster_centers_pca = kmeans_pca.cluster_centers_
plt.scatter(cluster_centers_pca[:,0], cluster_centers_pca[:,1], 
           c='red', marker='X', s=200, edgecolors='black', linewidth=2,
           label='Cluster Centers')

plt.xlabel(f'PC1 ({explained_variance[0]*100:.1f}% variance)')
plt.ylabel(f'PC2 ({explained_variance[1]*100:.1f}% variance)')
plt.title('K-Means Clusters with Centers (PCA Space)')
plt.colorbar(scatter, label='Cluster')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
sil_pca = silhouette_score(X_pca, clusters_pca)
dbi_pca = davies_bouldin_score(X_pca, clusters_pca)
chi_pca = calinski_harabasz_score(X_pca, clusters_pca)

print("PCA Data Metrics")
print("Silhouette:", sil_pca)
print("DBI:", dbi_pca)
print("CHI:", chi_pca)
```

```python
import pandas as pd

comparison_df = pd.DataFrame({
    'Metric': ['Silhouette Score', 'Davies-Bouldin Index', 'Calinski-Harabasz Index'],
    'Original Data': [sil, dbi, chi],
    'PCA-Reduced Data': [sil_pca, dbi_pca, chi_pca],
    'Improvement': [f"{(sil_pca-sil)/sil*100:.1f}%", 
                    f"{(dbi-dbi_pca)/dbi*100:.1f}%",
                    f"{(chi_pca-chi)/chi*100:.1f}%"]
})

print("\n" + "="*60)
print("PERFORMANCE COMPARISON: ORIGINAL vs PCA-REDUCED DATA")
print("="*60)
print(comparison_df.to_string(index=False))
```

Applying PCA before K-means clustering substantially improved clustering performance. The first two principal components retained approximately 70.56% of the dataset variance, preserving most of the important information while reducing dimensionality. After PCA transformation, the silhouette score increased from 0.277 to 0.454, the Davies–Bouldin Index decreased from 1.205 to 0.728, and the Calinski–Harabasz Index increased from 104.16 to 246.74. These results indicate that PCA enhanced cluster compactness and separation by reducing noise and emphasizing the dominant structure within the dataset.

```python
# COMPARISON WITH TRUE LABELS 
from sklearn.metrics import confusion_matrix, classification_report, adjusted_rand_score, normalized_mutual_info_score

# Get actual labels
actual_labels = df['status'].values

# Map each cluster to majority class using numpy (most compatible)
cluster_to_label = {}
for i in range(k_optimal):
    mask = (clusters_pca == i)
    if mask.sum() > 0:
        cluster_labels = actual_labels[mask]
        # Find most frequent label using numpy
        unique, counts = np.unique(cluster_labels, return_counts=True)
        majority_class = unique[np.argmax(counts)]
        cluster_to_label[i] = majority_class

# Predict labels based on cluster assignment
predicted_labels = np.array([cluster_to_label[label] for label in clusters_pca])

# Calculate accuracy
accuracy = (predicted_labels == actual_labels).mean()
print(f"\n{'='*60}")
print("COMPARISON WITH GROUND TRUTH LABELS")
print("="*60)
print(f"Clustering Accuracy (after label matching): {accuracy:.4f} ({accuracy*100:.2f}%)")

# Confusion Matrix
cm = confusion_matrix(actual_labels, predicted_labels)
print("\nConfusion Matrix:")
print("Rows: Actual (0=Healthy, 1=Parkinson's)")

# Classification report
print("\nClassification Report:")
print(classification_report(actual_labels, predicted_labels, 
                          target_names=['Healthy', "Parkinson's"]))

# Additional metrics
ari_score = adjusted_rand_score(actual_labels, clusters_pca)
nmi_score = normalized_mutual_info_score(actual_labels, clusters_pca)
print(f"\nAdjusted Rand Index (ARI): {ari_score:.4f}")
print(f"Normalized Mutual Info (NMI): {nmi_score:.4f}")
```

```python
print("\n" + "="*60)
print("FINAL CLUSTER COMPOSITION SUMMARY")
print("="*60)

for i in range(k_optimal):
    mask = (clusters_pca == i)
    parkinson_count = (actual_labels[mask] == 1).sum()
    healthy_count = (actual_labels[mask] == 0).sum()
    percentage = parkinson_count/mask.sum()*100 if mask.sum() > 0 else 0
    print(f"\nCluster {i}: {mask.sum()} samples")
    print(f"  • Healthy: {healthy_count}")
    print(f"  • Parkinson's: {parkinson_count}")
    print(f"  → {percentage:.1f}% Parkinson's patients")

print("\n CLINICAL INTERPRETATION:")
print("  The 3 clusters likely represent different disease stages:")
print("  • Cluster 0 (78 patients, 98.7% PD): High-confidence Parkinson's group")
print("  • Cluster 1 (11 patients, 100% PD): Small, potentially severe/advanced PD subgroup")
print("  • Cluster 2 (106 patients, 55.7% PD): Mixed/early-stage or milder PD group")
print("\n  This 3-cluster solution provides more granular insight than simple")
print("  binary classification, potentially capturing disease severity variation.")
```

**Hierarchial**

```python
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from scipy.spatial.distance import pdist
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score, confusion_matrix, classification_report
```

```python
# Determine Optimal Number of Clusters using Dendrogram

print("\n STEP 1: Determining Optimal Clusters using Dendrogram")

# Compute linkage matrix for dendrogram
linkage_matrix = linkage(X_scaled, method='ward')

# Plot dendrogram to find optimal k
plt.figure(figsize=(14, 7))
dendrogram(linkage_matrix, truncate_mode='lastp', p=30, 
           leaf_rotation=90., leaf_font_size=10., 
           show_contracted=True)
plt.axhline(y=25, color='r', linestyle='--', alpha=0.7, label='Potential cutoff')
plt.title('Hierarchical Clustering Dendrogram', fontsize=14, fontweight='bold')
plt.xlabel('Sample Index', fontsize=12)
plt.ylabel('Distance (Ward Linkage)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

print("\nBased on the dendrogram, recommended number of clusters: 3")
optimal_k_hier = 3
```

```python
# Fit Agglomerative Clustering
hier_clustering = AgglomerativeClustering(
    n_clusters=optimal_k_hier,
    linkage='ward',
    metric='euclidean'
)
hier_labels = hier_clustering.fit_predict(X_scaled)

print(f"\nModel Summary:")
print(f"  • Number of clusters: {optimal_k_hier}")
print(f"  • Linkage method: Ward")
print(f"  • Distance metric: Euclidean")
print(f"  • Number of samples: {len(hier_labels)}")

# Cluster distribution
print("\nCluster Distribution:")
unique, counts = np.unique(hier_labels, return_counts=True)
for i, count in zip(unique, counts):
    print(f"  • Cluster {i}: {count} samples ({count/len(hier_labels)*100:.1f}%)")
```

Cluster Distribution
Cluster 0: 142 samples (72.8%) → Predominantly Parkinson's (66.9%)

Cluster 1: 49 samples (25.1%) → 98% Parkinson's patients

Cluster 2: 4 samples (2.1%) → 100% Parkinson's (small, potentially severe subgroup)

```python
# Evaluation Metrics (Original Data)

sil_hier = silhouette_score(X_scaled, hier_labels)
dbi_hier = davies_bouldin_score(X_scaled, hier_labels)
chi_hier = calinski_harabasz_score(X_scaled, hier_labels)

print(f"  • Silhouette Score: {sil_hier:.4f}")
print(f"  • Davies-Bouldin Index: {dbi_hier:.4f}")
print(f"  • Calinski-Harabasz Index: {chi_hier:.2f}")
```

The PCA plot (first two components explaining 59.3% and 11.3% variance) shows that hierarchical clustering primarily separates a high-confidence Parkinson's subgroup (Cluster 1) from a larger mixed group (Cluster 0), while Cluster 2 represents a small, potentially severe case subgroup.

```python
# Comparison with Ground Truth Labels

# Map each cluster to majority class
cluster_to_label_hier = {}
for i in range(optimal_k_hier):
    mask = (hier_labels == i)
    if mask.sum() > 0:
        cluster_labels = actual_labels[mask]
        unique, counts = np.unique(cluster_labels, return_counts=True)
        majority_class = unique[np.argmax(counts)]
        cluster_to_label_hier[i] = majority_class

# Predict labels based on cluster assignment
predicted_labels_hier = np.array([cluster_to_label_hier[label] for label in hier_labels])

# Calculate accuracy
accuracy_hier = (predicted_labels_hier == actual_labels).mean()
print(f"\nClustering Accuracy (after label matching): {accuracy_hier:.4f} ({accuracy_hier*100:.2f}%)")

# Confusion Matrix
cm_hier = confusion_matrix(actual_labels, predicted_labels_hier)
print("\nConfusion Matrix:")
print("Rows: Actual (0=Healthy, 1=Parkinson's)")
print("Columns: Predicted")
print(cm_hier)

# Classification Report
print("\nClassification Report:")
print(classification_report(actual_labels, predicted_labels_hier, 
                          target_names=['Healthy', "Parkinson's"]))

# Additional metrics
ari_hier = adjusted_rand_score(actual_labels, hier_labels)
nmi_hier = normalized_mutual_info_score(actual_labels, hier_labels)
print(f"\nAdjusted Rand Index (ARI): {ari_hier:.4f}")
print(f"Normalized Mutual Info (NMI): {nmi_hier:.4f}")
```

The low ARI/NMI scores indicate that hierarchical clustering captures different groupings than the binary healthy/Parkinson's labels, which is expected since the method identifies 3 subgroups rather than just 2 disease states.

```python
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.decomposition import PCA

print("\nSTEP 1: DBSCAN parameter sweep to identify a stable eps")

try:
    actual_labels
except NameError:
    actual_labels = df['status'].values

try:
    X_pca
    explained_variance
except NameError:
    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X_scaled_df)
    explained_variance = pca.explained_variance_ratio_

eps_values = np.arange(0.3, 1.6, 0.1)
results = []
for eps in eps_values:
    dbscan_model = DBSCAN(eps=eps, min_samples=5, metric='euclidean')
    labels_dbscan = dbscan_model.fit_predict(X_scaled_df)
    n_clusters = len(set(labels_dbscan)) - (1 if -1 in labels_dbscan else 0)
    n_noise = np.sum(labels_dbscan == -1)
    sil_score = np.nan
    if n_clusters > 1:
        sil_score = silhouette_score(X_scaled_df, labels_dbscan)
    results.append((eps, n_clusters, n_noise, sil_score))

results_df = pd.DataFrame(results, columns=['eps', 'n_clusters', 'n_noise', 'silhouette'])
print(results_df.to_string(index=False))

best = results_df.loc[results_df['n_clusters'] > 1].sort_values(['silhouette', 'n_clusters'], ascending=[False, False]).head(1)
if not best.empty:
    best_eps = best['eps'].values[0]
else:
    best_eps = 0.8
    print("\nNo stable eps found with more than one cluster; defaulting to eps=0.8")

print(f"\nSelected eps for final DBSCAN: {best_eps}")

# Fit final DBSCAN model
final_dbscan = DBSCAN(eps=best_eps, min_samples=5, metric='euclidean')
final_labels = final_dbscan.fit_predict(X_scaled_df)

n_clusters_dbscan = len(set(final_labels)) - (1 if -1 in final_labels else 0)
noise_count = np.sum(final_labels == -1)
print(f"\nDBSCAN found {n_clusters_dbscan} clusters and {noise_count} noise points.")

if n_clusters_dbscan > 1:
    sil_dbscan = silhouette_score(X_scaled_df, final_labels)
    dbi_dbscan = davies_bouldin_score(X_scaled_df, final_labels)
    chi_dbscan = calinski_harabasz_score(X_scaled_df, final_labels)
    print(f"Silhouette Score: {sil_dbscan:.4f}")
    print(f"Davies-Bouldin Index: {dbi_dbscan:.4f}")
    print(f"Calinski-Harabasz Index: {chi_dbscan:.2f}")
else:
    sil_dbscan = dbi_dbscan = chi_dbscan = np.nan
    print("DBSCAN did not identify more than one cluster; metric scores are not defined.")

# Visualize DBSCAN in PCA space -
plt.figure(figsize=(8,6))
unique_labels = np.unique(final_labels)
palette = sns.color_palette('tab10', len(unique_labels))
for label, color in zip(unique_labels, palette):
    mask = final_labels == label
    label_name = 'Noise' if label == -1 else f'Cluster {label}'
    plt.scatter(X_pca[mask,0], X_pca[mask,1], c=[color], label=label_name, s=50, alpha=0.7, edgecolors='k')

plt.title('DBSCAN Clusters (PCA Projection)')
plt.xlabel(f'PC1 ({explained_variance[0]*100:.1f}% variance)')
plt.ylabel(f'PC2 ({explained_variance[1]*100:.1f}% variance)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Report cluster composition
print("\nDBSCAN cluster composition:")
for label in unique_labels:
    mask = final_labels == label
    label_name = 'Noise' if label == -1 else f'Cluster {label}'
    print(f"  • {label_name}: {mask.sum()} samples")

if n_clusters_dbscan > 0:
    print("\nClass composition within DBSCAN clusters:")
    for label in sorted(set(final_labels)):
        if label == -1:
            continue
        mask = final_labels == label
        parkinson_count = (actual_labels[mask] == 1).sum()
        healthy_count = (actual_labels[mask] == 0).sum()
        percentage = parkinson_count / mask.sum() * 100 if mask.sum() > 0 else 0
        print(f"    Cluster {label}: {mask.sum()} samples | Healthy={healthy_count} | Parkinson's={parkinson_count} | {percentage:.1f}% PD")
```

## DBSCAN Summary

- DBSCAN with `eps = 1.5` and `min_samples = 5` identified 5 clusters plus 150 noise points.
- The clustering quality was poor for this dataset: silhouette score was `-0.1755`, indicating that clusters are not well-separated.
- Most points were labeled as noise, suggesting that the dataset is not a good fit for DBSCAN without additional preprocessing or parameter tuning.
- Among the 5 clusters found, two were pure Parkinson's clusters (`100%` PD), one cluster was strongly Parkinson's (`70%` PD), and two clusters were predominantly healthy.

This result suggests DBSCAN is not the best clustering method for the current scaled Parkinson's dataset in its existing form.
